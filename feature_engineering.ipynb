```python
import json
import pandas as pd
import numpy as np
import os
```

```python
# Read the WRDS annual fundamentals CSV from the wrds_data directory
wrds_annual_fundamentals_path = os.path.join("wrds_data", "wrds_annual_fundamentals.csv")
wrds_annual_fundamentals_df = pd.read_csv(wrds_annual_fundamentals_path)
wrds_annual_fundamentals_df_original = wrds_annual_fundamentals_df.copy()
# By default, Jupyter notebook's display() may not allow horizontal scrolling for wide DataFrames.
# Use pd.set_option to allow horizontal scrolling in notebook output:
pd.set_option('display.width', 2000)
pd.set_option('display.max_columns', None)
display(wrds_annual_fundamentals_df.head())
```

```python
# Find columns where all values are null
null_columns = wrds_annual_fundamentals_df.columns[wrds_annual_fundamentals_df.isnull().all()].tolist()

# Print before column count
print("Number of columns before removal:", wrds_annual_fundamentals_df.shape[1])

# Remove all-null columns
wrds_annual_fundamentals_df = wrds_annual_fundamentals_df.drop(columns=null_columns)

# Print after column count
print("Number of columns after removal:", wrds_annual_fundamentals_df.shape[1])

# Print list of removed columns
print("All-null columns removed:", null_columns)

```

```python
# Print number of columns before dropping columns with >80% nulls
print("Number of columns before dropping >80% nulls:", wrds_annual_fundamentals_df.shape[1])

# Drop columns where more than 80% of the values are null
threshold = 0.8
cols_to_drop = wrds_annual_fundamentals_df.columns[wrds_annual_fundamentals_df.isnull().mean() > threshold].tolist()
wrds_annual_fundamentals_df = wrds_annual_fundamentals_df.drop(columns=cols_to_drop)

# Print number of columns after dropping
print("Number of columns after dropping >80% nulls:", wrds_annual_fundamentals_df.shape[1])
print("Columns dropped due to >80% nulls:", cols_to_drop)
```

```python
print("Remaining columns after cleaning:")
print(wrds_annual_fundamentals_df.columns.tolist())
```

```python
# Compare columns in the original and cleaned DataFrames to find dropped columns
dropped_columns = set(wrds_annual_fundamentals_df_original.columns) - set(wrds_annual_fundamentals_df.columns)
print("Columns dropped during cleaning:")
print(dropped_columns)
# for col in dropped_columns:
#     print(col)
```
